```python
import torch
from torchvision import datasets, transforms

# Define image transformations
transform = transforms.Compose([
    transforms.Resize((32, 32)),         # resize images to 32x32
    transforms.ToTensor(),               # convert image to tensor
    transforms.Normalize((0.5,), (0.5,)) # normalize pixel values
])

# Load CIFAR-10 dataset
train_data = datasets.CIFAR10(root='./data', train=True, download=True, transform=transform)
test_data = datasets.CIFAR10(root='./data', train=False, download=True, transform=transform)
```

```python
0 - airplane
1 - automobile
2 - bird
3 - cat
4 - deer
5 - dog
6 - frog
7 - horse
8 - ship
9 - truck
```

```python
# Extract data and targets
cat_dog_indices = [i for i, label in enumerate(train_data.targets) if label in [3, 5]]

# Filter dataset
filtered_data = [train_data.data[i] for i in cat_dog_indices]
filtered_labels = [1 if train_data.targets[i] == 3 else 0 for i in cat_dog_indices]
```

```python
from torch.utils.data import Dataset

class CatDogDataset(Dataset):
    def __init__(self, images, labels, transform=None):
        self.images = images
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        image = self.images[idx]
        label = self.labels[idx]

        # Convert to PIL Image before applying transform
        from PIL import Image
        image = Image.fromarray(image)

        if self.transform:
            image = self.transform(image)
        
        return image, label
```

```python
cat_dog_train = CatDogDataset(filtered_data, filtered_labels, transform=transform)
print("Total images:", len(cat_dog_train))
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(cat_dog_train, batch_size=32, shuffle=True)
```
